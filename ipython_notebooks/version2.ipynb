```python
import subprocess
import pandas as pd
import os

# In ra thư mục làm việc hiện tại để tham khảo
print("Current working directory:", os.getcwd())

# Đặt repo_dir là thư mục chứa Git repository của dự án.
# Ví dụ: nếu key của dự án là COPYOFTESTER, repository nằm tại /home/dataiku/dss/projects/COPYOFTESTER/
repo_dir = '/data/dataiku/datadir/jupyter-run/dku-workdirs/DKU_EXAM_DEVELOPER/version2bfc56896-dssuser_quka183741dd3eee'

# Kiểm tra xem repo_dir có tồn tại hay không và chứa thư mục .git
if not os.path.exists(repo_dir):
    print("Đường dẫn repo_dir không tồn tại:", repo_dir)
elif '.git' not in os.listdir(repo_dir):
    print("Thư mục repo_dir không chứa Git repository (không có thư mục .git):", repo_dir)
else:
    def get_git_log(cwd=None):
        """
        Chạy lệnh git log để lấy thông tin commit với định dạng:
          - %h: commit ID rút gọn
          - %s: Commit message
        Các trường được phân cách bởi dấu "|".
        """
        command = ['git', 'log', '--pretty=format:%h|%s']
        try:
            output = subprocess.check_output(command, universal_newlines=True, cwd=cwd)
        except subprocess.CalledProcessError as e:
            print("Lỗi khi chạy lệnh git log:", e)
            return None
        return output

    # Lấy kết quả của git log từ thư mục repo_dir
    log_output = get_git_log(cwd=repo_dir)

    if log_output:
        # Tách kết quả theo từng dòng (mỗi dòng là một commit)
        commit_lines = log_output.strip().split('\n')
        commit_data = []
        for line in commit_lines:
            # Mỗi dòng theo định dạng: commitID|commit message
            parts = line.split('|')
            if len(parts) >= 2:
                commit_hash = parts[0]
                commit_message = parts[1]
                commit_data.append((commit_hash, commit_message))
        # Tạo DataFrame với 2 cột: Commit ID và Message
        df = pd.DataFrame(commit_data, columns=['Commit ID', 'Message'])
        print(df)
    else:
        print("Không lấy được thông tin từ git log.")
```

```python
import os

# Hàm tìm thư mục chứa Git repository (có thư mục .git) bắt đầu từ một đường dẫn nhất định.
def find_git_repo(start_path):
    current_path = start_path
    while True:
        # Kiểm tra xem trong current_path có thư mục .git hay không
        if os.path.isdir(os.path.join(current_path, '.git')):
            return current_path
        # Lấy thư mục cha của current_path
        parent_path = os.path.dirname(current_path)
        # Nếu thư mục cha bằng với current_path nghĩa là đã đạt đến gốc hệ thống
        if parent_path == current_path:
            return None
        current_path = parent_path

# In ra thư mục làm việc hiện tại
current_dir = os.getcwd()
print("Thư mục làm việc hiện tại:", current_dir)

# Tìm thư mục chứa Git repository (version control) từ thư mục hiện tại
vc_dir = find_git_repo(current_dir)
if vc_dir:
    print("Thư mục version control (Git repository) được tìm thấy tại:", vc_dir)
    # In ra danh sách các file và thư mục trong thư mục version control
    print("Nội dung của thư mục version control:", os.listdir(vc_dir))
else:
    print("Không tìm thấy thư mục version control (Git repository) từ thư mục hiện tại.")
```
