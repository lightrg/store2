```python
# Lấy thông tin về Git repository của dự án
project_git = project.get_project_git()
```

```python
# Lấy danh sách các commit
commits = project_git.log()
```

```python
# Hiển thị thông tin các commit
for commit in commits:
    print("Commit ID:", commit['commit'])
    print("Tác giả:", commit['author'])
    print("Ngày giờ:", commit['date'])
    print("Thông điệp:", commit['message'])
    print("-" * 40)
```
