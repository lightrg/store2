```python
# Lấy thông tin về Git repository của dự án
project_git = project.get_project_git()
```

```python
# Lấy danh sách các commit
commit_log= project_git.log()
```

```python
# Phân tách chuỗi thành các dòng
lines = commit_log.split('\n')

# Biến lưu trữ thông tin commit hiện tại
current_commit = {}

# Danh sách chứa tất cả các commit
commits = []

# Duyệt qua từng dòng để trích xuất thông tin
for line in lines:
    if line.startswith('commit '):
        # Nếu đã có thông tin về commit trước đó, thêm nó vào danh sách
        if current_commit:
            commits.append(current_commit)
            current_commit = {}
        # Lưu trữ mã băm của commit
        current_commit['commit'] = line.split(' ')[1]
    elif line.startswith('Author:'):
        # Lưu trữ thông tin tác giả
        current_commit['author'] = line[len('Author: '):]
    elif line.startswith('Date:'):
        # Lưu trữ ngày tháng
        current_commit['date'] = line[len('Date: '):]
    else:
        # Giả định rằng các dòng còn lại là thông điệp commit
        if 'message' not in current_commit:
            current_commit['message'] = line.strip()
        else:
            current_commit['message'] += ' ' + line.strip()

# Thêm commit cuối cùng vào danh sách
if current_commit:
    commits.append(current_commit)

# Hiển thị thông tin các commit
for commit in commits:
    print("Commit ID:", commit.get('commit', 'N/A'))
    print("Tác giả:", commit.get('author', 'N/A'))
    print("Ngày giờ:", commit.get('date', 'N/A'))
    print("Thông điệp:", commit.get('message', 'N/A'))
    print("-" * 40)
```
