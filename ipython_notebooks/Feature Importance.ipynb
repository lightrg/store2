```python
import dataiku
import pandas as pd
from sklearn.feature_selection import SelectKBest, chi2
```

```python
# Read the dataset as a Pandas dataframe in memory
# Note: here, we only read the first 100K rows. Other sampling options are available
dataset_heart_disease_imputed_values = dataiku.Dataset("heart_disease_imputed_values")
df = dataset_heart_disease_imputed_values.get_dataset()
```

```python
X, y = df.drop(['location', 'at_risk', 'risk_level'], axis=1), df.at_risk
```

```python
best_features = SelectKBest(chi2, k=10).fit(X, y).get_support(indices=True)
```

```python
best_features_df = X.iloc[:, best_features]
best_features_df
```
