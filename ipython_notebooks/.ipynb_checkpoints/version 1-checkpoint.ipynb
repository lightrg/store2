```python
import csv
import os
from dataiku import api_client
```

```python
from dataiku import api_client

# Khởi tạo client để kết nối với Dataiku
client = api_client()

# Lấy đối tượng dự án bằng khóa dự án (project_key)
project = client.get_project('DKU_EXAM_DEVELOPER')

# Lấy đối tượng Git của dự án
project_git = project.get_project_git()

# Lấy lịch sử commit dưới dạng chuỗi văn bản
commit_log = project_git.log()

# Phân tách chuỗi thành các dòng
lines = commit_log

```

```python
commit_log
```

```python

# Create the 'output' directory if it doesn't exist
output_folder = "output"
os.makedirs(output_folder, exist_ok=True)

# Parse and structure commit information
commits = []

for commit in commit_log.get("commits", []):  # Assuming the log contains a dictionary with "commits"
    commit_entry = {
        "hash": commit.get("hash"),
        "author": commit.get("author"),
        "date": commit.get("date"),
        "message": commit.get("message")
    }
    commits.append(commit_entry)
csv_filename = os.path.join(output_folder, "commit_log.csv")
with open(csv_filename, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=["hash", "author", "date", "message"])
    writer.writeheader()
    writer.writerows(commits)
 

print(f"Commit log successfully saved to {csv_filename}.")
```

```python
import json
json_filename = os.path.join(output_folder, "commit_log.json")

with open(json_filename, mode='w', encoding='utf-8') as file:
    json.dump(commits, file, indent=4, ensure_ascii=False)

print(f"Commit log successfully saved to {json_filename}.")
```
